#### Exercise 1

Read in the simulated gene expression dataset found in `gene_data.csv`. Then fit a multiple linear regression model and compute the RMSE for this model.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
df = pd.read_csv("../data/gene_data.csv")

y = df["Trait"]
x = df.drop(columns="Trait")

full_model = LinearRegression()
full_model.fit(x,y)
full_pred = full_model.predict(x)

print("RMSE:", root_mean_squared_error(full_pred, y))
```

#### Exercise 2

##### Part 1

Split the data into training and test datasets using 20% of the data for testing and 80% for training.

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

##### Part 2

Fit the multiple linear regression model to the training data and then compute the test RMSE for the test dataset.

```python
train_model = LinearRegression()
train_model.fit(x_train, y_train)

test_pred = train_model.predict(x_test)
train_pred = train_model.predict(x_train)

print("Test RMSE:", root_mean_squared_error(test_pred, y_test))
print("Train RMSE:", root_mean_squared_error(train_pred, y_train))
```

##### Part 3

What do you observed about the train RMSE vs the test RMSE? Why do you think this is happening?

The training model is almost perfect because of it being very low (amost 0) but the error increases in a larger degree in the testing model. This suggests the model maybe overfitting the training data and does not generalize to unseen data.

##### Part 1

Use the `StandardScaler` class to compute means and standard deviations for each predictor in the training dataset.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(x_train)
```

##### Part 2

Standardize the predictors in the test dataset using the means and standard deviations computed in Part 1 with the `StandardScaler` class.

```python
x_test_scaled = scaler.transform(x_test)
```

#### Exercise 4

##### Part 1

Utilize the `LassoCV` class to use cross-validation to find the best value for $\alpha$ in the LASSO regression model and then compute the RMSE for the selected model.

```python
from sklearn.linear_model import LassoCV

x_train_scaled = scaler.transform(x_train)

lassoCV_model = LassoCV(cv=10)
lassoCV_model.fit(x_train_scaled, y_train)

lassoCV_model_pred = lassoCV_model.predict(x_test_scaled)

print("Selected alpha:", lassoCV_model.alpha_)
print("RMSE:", root_mean_squared_error(lassoCV_model_pred, y_test))

```

##### Part 2

In your own words, explain what `LassoCV` is doing. Then explain the difference between train, validation, and test datasets.

LassoCV uses cross validation to find the best alpha value for a LASSO regression model. When cv=10, the training data is divided into 10 folds. The model trains on 9 folds and uses the remaining fold as a validation set. This process is repeated 10 times so that each fold is used as the validation set once. LassoCV then compares the results from all the folds and selects the alpha value that performs best on average. The training dataset is used to train the model, the validation dataset is used to help choose the best alpha value, and the test dataset is used at the end to evaluate how well the final model performs on new unseen data.

##### Part 3

What is overfitting? How does LASSO regression utilize regularization to prevent overfitting?

overfitting happens when the model memorizes the training data and does not provide good predictions on the new data. To prevent that, we use LASSO regularization by reducing the size of the coefficient and removing any less important variables which makes the model simpler and better at generalizing new data.

##### Part 4

What exactly is the LASSO objective function designed to do and how does it prevent overfitting?

The LASSO objective function is designed to find the best-fitting regression model while also keeping the model simple. It does this by adding a penalty for large coefficients. As a result, LASSO shrinks less important coefficients toward zero and may even remove some variables completely. This helps prevent overfitting because the model focuses on the most important predictors instead of trying to memorize the training data.
